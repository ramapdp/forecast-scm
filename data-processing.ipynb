```python
import pandas as pd
import numpy as np

def load_data(file_path):
    """
    Load data from a CSV file into a pandas DataFrame.

    Parameters:
    file_path (str): The path to the CSV file.

    Returns:
    pd.DataFrame: A DataFrame containing the loaded data.
    """
    try:
        data = pd.read_csv(file_path, sep=';')
        return data
    except Exception as e:
        print(f"Error loading data: {e}")
        return None
```

```python
dataset = load_data('dataset/dataset.csv')

print("Data loaded successfully. Here are the first few rows:")
print(dataset.head())
```

### Make sure complete data from jan 24 - dec 25

```python
dataset['Tanggal'] = pd.to_datetime(dataset['Tanggal'])

start_date = dataset['Tanggal'].min()
end_date = dataset['Tanggal'].max()

print(f"Checking data from {start_date.strftime('%d %b %Y')} to {end_date.strftime('%d %b %Y')}")
print("-" * 40)

expected_dates = pd.date_range(start=start_date, end=end_date)

actual_dates = dataset['Tanggal'].dropna().unique()

missing_dates = expected_dates.difference(actual_dates)

if missing_dates.empty:
    print("Your daily data is complete! No missing dates.")
else:
    print(f"You are missing data for {len(missing_dates)} days.")
    print("Missing dates:")
    print(missing_dates.strftime('%d %b %Y').tolist())
```

### Check duplicated data

```python
total_duplicates = dataset.duplicated().sum()
print(f"Total duplicate rows: {total_duplicates}")
print("-" * 40)

# (keep=False ensures we see the original row AND its duplicates)
duplicates = dataset[dataset.duplicated(keep=False)]

if total_duplicates > 0:
    print("Here is a peek at the duplicate rows:")
    print(duplicates.sort_values(by=['Tanggal', 'Kode Barang', 'Nama Cabang']).head(10))
else:
    print("Your dataset is clean! No identical duplicate rows found.")
```

```python
import normalize_items
import build_panel
import calendar_features
import prepare_forecast_data

normalized = normalize_items.load_and_normalize()
print(f"{len(normalized)} rows after item-code normalization and reaggregation")
```

```python
panel = build_panel.build_dense_panel(normalized)
print(f"{len(panel)} rows after dense daily zero-fill ({panel[build_panel.PAIR_COLS].drop_duplicates().shape[0]} item-branch pairs)")
```

```python
panel = build_panel.filter_min_history(panel)
print(f"{len(panel)} rows after minimum-history filter ({panel[build_panel.PAIR_COLS].drop_duplicates().shape[0]} item-branch pairs)")
```

```python
featured = prepare_forecast_data.add_targets(panel)
featured = prepare_forecast_data.add_lag_features(featured)
featured = prepare_forecast_data.add_rolling_features(featured)
featured = calendar_features.add_calendar_features(featured)
branch_stats = prepare_forecast_data.compute_branch_stats(featured)
featured = prepare_forecast_data.apply_branch_stats(featured, branch_stats)
featured = prepare_forecast_data.add_branch_age_days(featured)
print(f"{len(featured)} rows, {len(featured.columns)} columns after feature engineering")
```

```python
# Row-count sanity check
expected_rows = panel[build_panel.PAIR_COLS].drop_duplicates().shape[0]
assert len(featured) == len(panel), "feature engineering must not change row count"
print("Row count sanity check passed.")
```

```python
# No duplicate (pair, date) rows
dup_count = featured.duplicated(subset=build_panel.PAIR_COLS + ["Tanggal"]).sum()
assert dup_count == 0, f"found {dup_count} duplicate (pair, date) rows"
print("No duplicate (pair, date) rows.")
```

```python
# No negative quantities
assert (featured["Kuantitas"] >= 0).all(), "found negative Kuantitas values"
print("No negative Kuantitas values.")
```

```python
# Ramadan/Eid flags spot-check against verified dates
ramadan_2024_start = featured[featured["Tanggal"] == "2024-03-11"]
assert ramadan_2024_start["is_ramadan"].all(), "2024-03-11 should be flagged as Ramadan"
eid_fitr_2024 = featured[featured["Tanggal"] == "2024-04-10"]
assert eid_fitr_2024["is_eid_al_fitr"].all(), "2024-04-10 should be flagged as Eid al-Fitr"
print("Ramadan/Eid flag spot-check passed.")
```

```python
# Leakage check: a sampled row's lag/rolling features only reference strictly earlier dates
sample_pair = featured[build_panel.PAIR_COLS].drop_duplicates().iloc[0]
sample_series = featured[
    (featured["Kode Barang"] == sample_pair["Kode Barang"])
    & (featured["Nama Cabang"] == sample_pair["Nama Cabang"])
].sort_values("Tanggal").reset_index(drop=True)
if len(sample_series) > 28:
    row = sample_series.iloc[28]  # far enough in to have full lag history
    assert row["lag_1"] == sample_series.iloc[27]["Kuantitas"], "lag_1 must equal the prior day's Kuantitas"
    assert row["roll_mean_7"] == sample_series.iloc[21:28]["Kuantitas"].mean(), "roll_mean_7 must exclude today's own Kuantitas"
    print("Leakage spot-check passed: lag_1 and roll_mean_7 reference only strictly earlier dates.")
else:
    print("Sampled pair too short for a 28-day leakage check; pick a longer-history pair if this happens.")
```
